# **DBSCAN Algorithm**
### **Density-Based Spatial Clustering of Applications with Noise**
**DBSCAN** is a density-based clustering algorithm used in unsupervised machine learning to group together points that are closely packed (dense regions) and mark points that lie alone in low-density regions as outliers (noise).

## **Step 1: Import required libraries**

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons, make_blobs
```

### **Step 2: Generate random sample dataset 1**

```python
# Generate sample data
# Option 1: Two interleaving half circles (moons)
X, _ = make_moons(n_samples=300, noise=0.1)
```

```python
# Plot the sample data
plt.scatter(X[:, 0], X[:, 1])
plt.title("Sample Data")
plt.show()
```

## **Step 3: Apply DBSCAN Algorithm**

```python
# Apply DBSCAN
# Parameters: eps = 0.2, min_samples = 5
dbscanX = DBSCAN(eps=0.2, min_samples=5)
labelsX = dbscanX.fit_predict(X)
```

```python
# Plot the results
plt.scatter(X[:, 0], X[:, 1], c=labelsX, cmap='viridis')
plt.title("DBSCAN Clustering")
plt.show()

print("\n")
print("-----"*10)
print("\n")

# Output the labels
print("Cluster Labels:")
print(labelsX)
```

## **Step 4: Generate random sample dataset 2**

```python
# Option 2: Blob clusters
Y, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
```

```python
# Plot the sample data
plt.scatter(Y[:, 0], Y[:, 1])
plt.title("Sample Data")
plt.show()
```

## **Step 3: Apply DBSCAN Algorithm**

```python
# Apply DBSCAN
# Parameters: eps = 0.2, min_samples = 5
dbscanY = DBSCAN(eps=0.2, min_samples=5)
labelsY = dbscanY.fit_predict(Y)
```

```python
# Plot the results
plt.scatter(Y[:, 0], Y[:, 1], c=labelsY, cmap='viridis')
plt.title("DBSCAN Clustering")
plt.show()

print("\n")
print("-----"*10)
print("\n")

# Output the labels
print("Cluster Labels:")
print(labelsY)
```
